```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Order_ID": [101, 102, 103, 104],
    "Date": [
        "2024-01-05",
        "2024-01-08",
        "2024-02-10",
        "2024-03-15"
    ],
    "Month": ["Jan", "Jan", "Feb", "Mar"],
    "Product": ["Laptop", "Mobile", "Headphone", "Laptop"],
    "Category": ["Electronics", "Electronics", "Accessories", "Electronics"],
    "Units_Sold": [5, 10, 15, 7],
    "Unit_Price": [50000, 20000, 3000, 52000]
}

df = pd.DataFrame(data)
```

```python
print("Initial Data")
print(df)
print("\nInfo")
print(df.info())
```

```python
# Step 3 — Data Cleaning
print("\nMissing Values")
print(df.isnull().sum())
```

```python
# Remove duplicates
df.drop_duplicates(inplace=True)
```

```python
# Step 4 — Feature Engineering
df['Total_Sales'] = df['Units_Sold'] * df['Unit_Price']
df['Date'] = pd.to_datetime(df['Date'])
print("\nAfter Feature Engineering")
print(df)
```

```python
# Step 5 — EDA
print("\nStatistics")
print(df.describe())
```

```python
# Group by Category and sum Total_Sales
sales_array = np.array(df['Total_Sales'])
print("\nTotal Revenue:", np.sum(sales_array))
print("Average Order Value:", np.mean(sales_array))

```

```python
# Step 6 — Product-wise Analysis
product_sales = df.groupby('Product')['Total_Sales'].sum().sort_values(ascending=False)
print("\nProduct-wise Sales")
print(product_sales)
```

```python
# Step 7 — Category-wise Analysis
category_sales = df.groupby('Category')['Total_Sales'].sum()
print("\nCategory-wise Sales")
print(category_sales)
```

```python
# Step 8 — Monthly Trend
monthly_sales = df.groupby('Month')['Total_Sales'].sum()
plt.plot(monthly_sales.index, monthly_sales.values)
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.show()
```

```python
# Step 9 — Visualization

# Bar Chart
product_sales.plot(kind='bar')
plt.title('Product-wise Revenue')
plt.xlabel('Product')
plt.ylabel('Revenue')
plt.show()
```

```python
# Pie Chart
category_sales.plot(kind='pie', autopct='%1.1f%%')
plt.ylabel('')
plt.title('Category Contribution')
plt.show()
```

```python
# scatter plot
plt.scatter(df['Units_Sold'], df['Total_Sales'])
plt.title('Units Sold vs Total Sales')
plt.xlabel('Units Sold')
plt.ylabel('Total Sales')
plt.show()
```

```python
# boxplot
plt.boxplot(df['Total_Sales'])
plt.title('Boxplot of Total Sales')
plt.show()
```
